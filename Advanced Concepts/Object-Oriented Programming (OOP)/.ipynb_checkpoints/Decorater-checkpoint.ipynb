# Decoraters

It is the use for design the function

```python
class employee: 
    def __init__(self,name,salary):
        self.name = name 
        self.salary = salary 

    def greet(fx):
        def mfx(*args,**kwargs): 
            print("Hello !")
            fx(*args,**kwargs)
            print("Good Bye !")
        return mfx
        
    @greet
    def show_salary(self): 
        print(f"The Name Of Employye is {self.name} ")
        print(f"His Salary are {self.salary}")

emp1 = employee("swapnil",4000000)
emp1.show_salary()
```

```python
emp2 = employee("Avinash",5000000)
emp2.show_salary()
```

```python
emp3 = employee("Vijay",6000000)
emp3.show_salary()
```

```python
emp4 = employee("Vivek",7000000)
emp4.show_salary()
```

```python
class student:
    def __init__(self,name,marks):
        self.name = name
        self.marks = marks
        
    def greet(fx):
        def mfx(*args,**kwargs):
            print("Hello Student !")
            fx(*args,**kwargs)
            print("Thanks For Coming !")
        return mfx 

    @greet 
    def show_marks(self):
        print(f"The Student name is {self.name} ")
        print(f"His Marks are obtained {self.marks}%")

std1 = student("Swapnil Pisal",79.0)
std1.show_marks()
           
        
```

```python
std2 = student("Avinash Veer",85.5)
std2.show_marks()
```

```python
std3 = student("Vijay Jadhav",88.9)
std3.show_marks()
```

```python
std4 = student("Vivek Jadhav",91.9)
std4.show_marks()
```
