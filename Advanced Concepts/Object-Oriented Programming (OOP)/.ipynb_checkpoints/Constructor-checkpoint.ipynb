# Constructor

In python using __init__ method to create the constructer

```python
# Create the student class
```

```python
# class it is blueprint
class student:
    def __init__(self,name,age,marks):
        self.name = name 
        self.age = age 
        self.marks = marks 

    def info(self):
        print(f"The student name is {self.name} his age is {self.age} and his marks are obtained {self.marks} ")
```

```python
# objects
std1 = student("Swapnil Pisal",21,4000000)
std1.info()
```

```python
# use multiple object
std2 = student("Vivek Jadhav",21,500000)
std2.info()
```

```python
std3 = student("Vijay Jadhav",21,500000)
std3.info()
```

```python
std3 = student("Avi veer",21,600000)
std3.info()
```

```python
std4 = student("Tushar Jadhav",23,500000)
std4.info()
```

```python
class employee:
    def __init__(self,fname,lname,salary,experience): 
        self.fname = fname
        self.lname = lname
        self.salary = salary
        self.experience = experience

    def showdetail(self):
        print(f"The Employee Name is {self.fname} and his surname are {self.lname}")
        print(f"His Salary are {self.salary}rs")
        print(f"Salary per year {self.salary/self.experience}")
        
```

```python
emp1 = employee("Swapnil","Pisal",40000000,5)
emp1.showdetail()
```

```python
emp2 = employee("Avi","Patil",60000,2)
emp2.showdetail()
```
