```python
# 1) Student Example
```

```python
# class
# it is the blue print
class students:
    name = "Swapnil" 
    marks = 97

    def showdetail(self):
        print(f"The name of student is {self.name} and marks are obtained {self.marks} ")

# object 
# It use the multiple object using the same class
a = students()
a.showdetail()
print(a.name)
```

```python
b = students()
b.name = "Vivek"
b.marks = 56
b.showdetail()
```

```python
c=students()
c.showdetail()
```

```python
class employee: 
    name = "Ram Shinde"
    age = 34
    salary = 340000

    def show(self):
        print(f"The employye name is {self.name} his age is {self.age} and his monthly salary are {self.salary}rs ")

emp1 = employee()
emp1.show()
```

```python
emp2 = employee()
emp2.name = "Swapnil Pisal"
emp2.show()
```
