```python
class student: 
    def __init__(self,marks): 
        self.marks = marks 

    def info(self):
        print(f"The marks are {self.marks}")

    @property 
    def ten_marks(self): 
        return 10*self.marks

    @ten_marks.setter
    def ten_marks(self,new_marks): 
        self.marks = new_marks/10

a = student(10)
a.ten_marks = 90
# print(a.ten_marks())
a.info()
```
